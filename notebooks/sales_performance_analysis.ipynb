### 1. Import Libraries & Configure Environment

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization setup
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
print("Libraries successfully imported!")
```

### 2. Ingest and Inspect Sales Dataset

```python
data_path = os.path.join("..", "data", "cleaned_sales_data.csv")
df = pd.read_csv(data_path)
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
df["Ship_Date"] = pd.to_datetime(df["Ship_Date"])

print(f"Dataset Shape: {df.shape[0]:,} rows x {df.shape[1]} columns")
display(df.head(3))
```

### 3. Executive KPI Computation

```python
total_rev = df["Net_Sales"].sum()
total_prof = df["Profit"].sum()
margin = (total_prof / total_rev) * 100
total_orders = df["Order_ID"].nunique()
unique_cust = df["Customer_ID"].nunique()
aov = total_rev / total_orders

print("=" * 45)
print("         EXECUTIVE SALES SCORECARD")
print("=" * 45)
print(f"Total Net Revenue    : ${total_rev:,.2f}")
print(f"Gross Operating Profit: ${total_prof:,.2f}")
print(f"Consolidated Margin  : {margin:.2f}%")
print(f"Total Orders Fulfilled: {total_orders:,}")
print(f"Active Customer Base : {unique_cust:,}")
print(f"Average Order Value  : ${aov:,.2f}")
print("=" * 45)
```

### 4. Monthly Trend Analysis (Revenue & Profit Dynamics)

```python
monthly = df.groupby("Order_YearMonth").agg({
    "Net_Sales": "sum",
    "Profit": "sum"
}).reset_index()

plt.figure(figsize=(14, 5))
plt.plot(monthly["Order_YearMonth"], monthly["Net_Sales"] / 1000, marker="o", color="#2563EB", label="Revenue ($k)")
plt.plot(monthly["Order_YearMonth"], monthly["Profit"] / 1000, marker="s", color="#10B981", label="Profit ($k)")
plt.xticks(rotation=45, ha="right")
plt.title("Longitudinal Monthly Commercial Performance (2023 - 2025)", fontsize=14, fontweight="bold")
plt.ylabel("Amount ($k USD)")
plt.legend()
plt.tight_layout()
plt.show()
```

### 5. Category & Subcategory Profitability Breakdown

```python
subcat = df.groupby(["Category", "Sub_Category"]).agg({
    "Net_Sales": "sum",
    "Profit": "sum"
}).reset_index()
subcat["Margin_Pct"] = (subcat["Profit"] / subcat["Net_Sales"]) * 100
subcat = subcat.sort_values("Margin_Pct", ascending=False)

plt.figure(figsize=(12, 6))
colors = ["#EF4444" if m < 10 else "#10B981" for m in subcat["Margin_Pct"]]
plt.barh(subcat["Sub_Category"][::-1], subcat["Margin_Pct"][::-1], color=colors[::-1])
plt.axvline(20, color="gray", linestyle="--", label="Target Margin (20%)")
plt.title("Subcategory Margin Ranking (Identifying Loss Leaders)", fontsize=14, fontweight="bold")
plt.xlabel("Profit Margin (%)")
plt.legend()
plt.tight_layout()
plt.show()
```

### 6. Regional Performance Comparison

```python
reg = df.groupby("Region").agg({
    "Net_Sales": "sum",
    "Profit": "sum",
    "Order_ID": "nunique"
}).reset_index()
reg["Profit_Margin"] = (reg["Profit"] / reg["Net_Sales"]) * 100
display(reg.sort_values("Net_Sales", ascending=False))

reg.plot(x="Region", y=["Net_Sales", "Profit"], kind="bar", figsize=(10, 5), color=["#3B82F6", "#10B981"])
plt.title("Regional Sales vs Operating Profit", fontsize=14, fontweight="bold")
plt.ylabel("USD ($)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
